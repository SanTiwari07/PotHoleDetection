## 1. Detect potholes in an image

```python
import urllib.request, os
from ultralytics import YOLO

MODEL_PATH = "assets/models/pothole_yolov8.pt"
if not os.path.exists(MODEL_PATH):
    os.makedirs(os.path.dirname(MODEL_PATH), exist_ok=True)
    urllib.request.urlretrieve(
        "https://github.com/SanTiwari07/PotHoleDetection/releases/download/v1.0/pothole_yolov8.pt", MODEL_PATH)

model = YOLO(MODEL_PATH)
results = model.predict("demo/examples/road_1.jpg", conf=0.25)
results[0].show()
```
